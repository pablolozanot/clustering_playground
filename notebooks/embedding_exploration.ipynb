```python
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df_sample = pd.read_parquet("../data/processed/complaints_50k.parquet")
```

```python
texts = df_sample[
    "Consumer complaint narrative"
].tolist()
```

```python
model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)
```

```python
## Generate embeddings for the texts
texts = (
    df_sample["Consumer complaint narrative"]
    .dropna()
    .astype(str)
    .tolist()
)
sample_texts = texts[:2000]
embeddings = model.encode(
    sample_texts,
    show_progress_bar=True
)
```

```python
## Now we compare against a query

#query = sample_texts[0]
#or
query = """
The bank froze my account after I received
an international transfer.
"""

print(query[:1000])
```

```python
query_embedding = model.encode([query])
similarities = cosine_similarity(
    query_embedding,
    embeddings
)

#find nearest
top_indices = similarities[0].argsort()[-6:][::-1]

for idx in top_indices:
    print("\n====================\n")
    print(f"Similarity: {similarities[0][idx]:.3f}")
    print(sample_texts[idx][:1500])
```
